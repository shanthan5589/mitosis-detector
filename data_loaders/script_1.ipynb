```python
import sqlite3

import pydicom
from pydicom.pixels import iter_pixels

import pandas as pd

import numpy as np

import os

from pathlib import Path
```

```python
# Create path for root directory. We haven't used "\" in Path("D:/") as it is interpreted as an escape character in python.
base_dir = Path("D:/") / "Mitosis WSI CCMCT"

# Create root directory folder.
os.makedirs(base_dir, exist_ok=True)
```

```python
# Creating Training Data Directory
training_data = base_dir / "training_data"
os.makedirs(training_data, exist_ok=True)

# Creating meta_data Directory to store all the tables in the Database
meta_data = base_dir / "meta_data"
os.makedirs(meta_data, exist_ok=True)
```

```python
# Database path, make sure to download this file seperately.
db_path = r"D:\Mitosis WSI CCMCT\sqlite_sequence.csv"

if not db_path.exists(): 
    raise FileNotFoundError("SQLite file missing")

# Connecting to the data base
conn = sqlite3.connect(db_path)

# Retrieving all the tables from the data base
query = "SELECT name FROM sqlite_master WHERE type='table';"
tables_list = pd.read_sql_query(query, conn)

# Storing all the tables in meta_data directory for easy access
for table_name in tables_list['name']:

    df = pd.read_sql_query(f"SELECT * FROM {table_name}", conn)
    file_path = base_dir / meta_data / f"{table_name}.csv"
    df.to_csv(file_path, index=False)

conn.close()
```

```python
# Creating 'mitotic' and 'non_mitotic' directory in base directory to segregate mitotic and non-mitotic
# cell pictures for the purpose of training.
os.makedirs(base_dir / 'mitotic', exist_ok=True)
os.makedirs(base_dir / 'non_mitotic', exist_ok=True)
```
