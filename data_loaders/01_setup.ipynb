# 1. Setup
Downloads the SQLite database, extracts metadata tables as CSVs,
and creates the directory structure for patch extraction.

```python
import sqlite3
import pandas as pd
import os
from pathlib import Path
```

```python
# Root directory — change this to match your setup
base_dir = Path("D:/Mitosis WSI CCMCT")

# Create directory structure
os.makedirs(base_dir / "training_data", exist_ok=True)
os.makedirs(base_dir / "meta_data", exist_ok=True)
os.makedirs(base_dir / "mitotic", exist_ok=True)
os.makedirs(base_dir / "non_mitotic", exist_ok=True)
```

```python
# Path to the SQLite database (downloaded separately from Kaggle)
db_path = base_dir / "training_data" / "MITOS_WSI_CCMCT_ODAEL_train_dcm.sqlite"

if not db_path.exists():
    raise FileNotFoundError(f"SQLite database not found at {db_path}")

# Extract all tables from the database and save as CSVs
conn = sqlite3.connect(str(db_path))

tables = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type="table";", conn)

for table_name in tables["name"]:
    df = pd.read_sql_query(f"SELECT * FROM {table_name}", conn)
    df.to_csv(base_dir / "meta_data" / f"{table_name}.csv", index=False)
    print(f"Saved {table_name}: {len(df)} rows")

conn.close()
```
